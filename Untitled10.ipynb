```python
# Final Assignment: Part 2 - Create Dashboard with Plotly and Dash
# Google Colab / Python script
#
# Expected CSV columns:
# Date, Year, Recession, Automobile_Sales, Vehicle_Type, Advertising_Expenditure,
# Consumer_Confidence, Price, Competition, GDP, Unemployment_Rate

import pandas as pd
import plotly.express as px
from dash import Dash, dcc, html, Input, Output

# ------------------------------------------------------------
# 1. Load data
# ------------------------------------------------------------
# Upload your CSV to Colab and change the filename below if needed.
DATA_FILE = "historical_automobile_sales.csv"

df = pd.read_csv(DATA_FILE)

# Convert date if present
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# ------------------------------------------------------------
# 2. Prepare common columns
# ------------------------------------------------------------
if "Year" not in df.columns and "Date" in df.columns:
    df["Year"] = df["Date"].dt.year

if "Recession" in df.columns:
    # Handle both 0/1 and text values
    recession_map = {
        0: "Non-Recession",
        1: "Recession",
        "0": "Non-Recession",
        "1": "Recession",
        "Non-Recession": "Non-Recession",
        "Recession": "Recession",
    }
    df["Economic_Condition"] = df["Recession"].map(recession_map).fillna(
        df["Recession"].astype(str)
    )

# ------------------------------------------------------------
# 3. Create Dash application
# ------------------------------------------------------------
app = Dash(__name__)
app.title = "Automobile Sales Dashboard"

years = sorted(df["Year"].dropna().unique()) if "Year" in df.columns else []
vehicle_types = sorted(df["Vehicle_Type"].dropna().unique()) if "Vehicle_Type" in df.columns else []

app.layout = html.Div(
    style={"fontFamily": "Arial", "padding": "20px"},
    children=[
        html.H1(
            "Automobile Sales Statistics Dashboard",
            style={"textAlign": "center"},
        ),

        html.Div(
            [
                html.Label("Select Report Type:"),
                dcc.Dropdown(
                    id="report_type",
                    options=[
                        {
                            "label": "Recession Period Statistics",
                            "value": "recession",
                        },
                        {
                            "label": "Yearly Statistics",
                            "value": "yearly",
                        },
                    ],
                    value="yearly",
                    clearable=False,
                    style={"width": "60%"},
                ),
            ],
            style={"marginBottom": "20px"},
        ),

        html.Div(
            [
                html.Label("Select Year:"),
                dcc.Dropdown(
                    id="year_dropdown",
                    options=[{"label": str(y), "value": y} for y in years],
                    value=years[-1] if years else None,
                    clearable=False,
                    style={"width": "60%"},
                ),
            ],
            id="year_container",
            style={"marginBottom": "20px"},
        ),

        html.Div(id="output_container"),

        dcc.Graph(id="sales_chart"),
    ],
)


# ------------------------------------------------------------
# 4. Dashboard callback
# ------------------------------------------------------------
@app.callback(
    Output("sales_chart", "figure"),
    Output("output_container", "children"),
    Input("report_type", "value"),
    Input("year_dropdown", "value"),
)
def update_dashboard(report_type, selected_year):

    if report_type == "recession":
        # Average automobile sales by vehicle type during
        # recession and non-recession periods.
        if "Economic_Condition" in df.columns and "Vehicle_Type" in df.columns:
            summary = (
                df.groupby(
                    ["Economic_Condition", "Vehicle_Type"],
                    as_index=False
                )["Automobile_Sales"]
                .mean()
            )

            fig = px.bar(
                summary,
                x="Economic_Condition",
                y="Automobile_Sales",
                color="Vehicle_Type",
                barmode="group",
                title="Average Vehicle-Wise Sales During Recession and Non-Recession",
                labels={
                    "Economic_Condition": "Economic Condition",
                    "Automobile_Sales": "Average Automobile Sales",
                    "Vehicle_Type": "Vehicle Type",
                },
            )

            message = html.H3(
                "Recession period automobile sales statistics",
                style={"textAlign": "center"},
            )

            return fig, message

    # --------------------------------------------------------
    # Yearly statistics
    # --------------------------------------------------------
    if selected_year is not None:
        yearly_df = df[df["Year"] == selected_year].copy()
    else:
        yearly_df = df.copy()

    if "Date" in yearly_df.columns:
        sales_by_date = (
            yearly_df.groupby("Date", as_index=False)["Automobile_Sales"]
            .mean()
        )

        fig = px.line(
            sales_by_date,
            x="Date",
            y="Automobile_Sales",
            markers=True,
            title=f"Automobile Sales Trend - {selected_year}",
            labels={
                "Date": "Date",
                "Automobile_Sales": "Average Automobile Sales",
            },
        )
    else:
        sales_by_vehicle = (
            yearly_df.groupby("Vehicle_Type", as_index=False)["Automobile_Sales"]
            .mean()
        )

        fig = px.bar(
            sales_by_vehicle,
            x="Vehicle_Type",
            y="Automobile_Sales",
            title=f"Average Automobile Sales by Vehicle Type - {selected_year}",
            labels={
                "Vehicle_Type": "Vehicle Type",
                "Automobile_Sales": "Average Automobile Sales",
            },
        )

    message = html.H3(
        f"Yearly automobile sales statistics for {selected_year}",
        style={"textAlign": "center"},
    )

    return fig, message


# ------------------------------------------------------------
# 5. Run the Dash application
# ------------------------------------------------------------
if __name__ == "__main__":
    app.run(debug=True)
```
