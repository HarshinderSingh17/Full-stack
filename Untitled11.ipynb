```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import folium

# Set style
sns.set_theme(style="whitegrid")

# Create sample dataset
data = {
    "City": ["Delhi", "Mumbai", "Bangalore", "Chennai", "Kolkata",
             "Hyderabad", "Pune", "Jaipur", "Ahmedabad", "Lucknow"],
    "Sales": [250, 300, 280, 220, 190, 270, 240, 150, 210, 170],
    "Profit": [50, 70, 65, 45, 35, 60, 55, 25, 40, 30],
    "Customers": [1200, 1500, 1400, 1100, 900, 1300, 1150, 700, 1000, 800],
    "Latitude": [28.6139, 19.0760, 12.9716, 13.0827, 22.5726,
                 17.3850, 18.5204, 26.9124, 23.0225, 26.8467],
    "Longitude": [77.2090, 72.8777, 77.5946, 80.2707, 88.3639,
                  78.4867, 73.8567, 75.7873, 72.5714, 80.9462]
}

df = pd.DataFrame(data)

df
plt.figure(figsize=(10, 5))

plt.plot(df["City"], df["Sales"], marker="o")

plt.title("Sales by City")
plt.xlabel("City")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.show()
plt.figure(figsize=(10, 5))

plt.bar(df["City"], df["Profit"])

plt.title("Profit by City")
plt.xlabel("City")
plt.ylabel("Profit")

plt.xticks(rotation=45)
plt.show()
plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=5)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
plt.figure(figsize=(8, 5))

plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
plt.figure(figsize=(10, 5))

sns.barplot(data=df, x="City", y="Sales")

plt.title("Sales by City")
plt.xticks(rotation=45)

plt.show()
plt.figure(figsize=(8, 5))

sns.boxplot(data=df[["Sales", "Profit", "Customers"]])

plt.title("Distribution of Sales, Profit and Customers")

plt.show()
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Sales", kde=True)

plt.title("Sales Distribution")

plt.show()
plt.figure(figsize=(8, 5))

sns.regplot(data=df, x="Sales", y="Profit")

plt.title("Relationship Between Sales and Profit")

plt.show()
plt.figure(figsize=(8, 6))

correlation = df[["Sales", "Profit", "Customers"]].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
m = folium.Map(
    location=[22.5, 78.9],
    zoom_start=5
)

m
for _, row in df.iterrows():

    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        popup=f"""
        <b>City:</b> {row['City']}<br>
        <b>Sales:</b> {row['Sales']}<br>
        <b>Profit:</b> {row['Profit']}<br>
        <b>Customers:</b> {row['Customers']}
        """,
        tooltip=row["City"]
    ).add_to(m)

m
m2 = folium.Map(
    location=[22.5, 78.9],
    zoom_start=5
)

for _, row in df.iterrows():

    folium.CircleMarker(
        location=[row["Latitude"], row["Longitude"]],
        radius=8,
        popup=f"{row['City']} - Sales: {row['Sales']}",
        tooltip=row["City"],
        fill=True
    ).add_to(m2)

m2
m2.save("india_sales_map.html")
```
